```python
import numpy as np
import random

class HospitalTriageEnv:
    def __init__(self):
        self.num_severity_states = 5
        self.num_congestion_states = 2
        self.action_space = [0, 1, 2]
        self.reset()

    def reset(self):
        self.current_severity = 4
        self.er_congestion = random.choice([0, 1])
        self.critical_allocated = False
        return (self.current_severity, self.er_congestion)

    def step(self, action):
        reward = 0
        done = False

        next_severity = self.current_severity - 1
        next_congestion = random.choice([0, 1])

        if action == 0:
            if self.current_severity >= 3 and not self.critical_allocated:
                reward = -25
        elif action == 1:
            if self.current_severity == 2:
                reward = 15
                self.critical_allocated = True
            else:
                reward = -4
        elif action == 2:
            if self.current_severity == 3:
                reward = 20
                self.critical_allocated = True
            else:
                reward = -10

        if self.er_congestion == 1 and action == 2 and self.current_severity < 3:
            reward -= 15

        if next_severity == 0:
            done = True
            if not self.critical_allocated:
                reward = -120
            else:
                reward += 60

        self.current_severity = next_severity
        self.er_congestion = next_congestion

        return (self.current_severity, self.er_congestion), reward, done

def run_hospital_mc_prediction(env, num_episodes=5000):
    V = np.zeros((env.num_severity_states, env.num_congestion_states))
    returns = {(s, c): [] for s in range(env.num_severity_states) for c in range(env.num_congestion_states)}

    for _ in range(num_episodes):
        state = env.reset()
        episode = []
        done = False

        while not done:
            action = 0
            next_state, reward, done = env.step(action)
            episode.append((state, action, reward))
            state = next_state

        G = 0
        visited_states = set()
        for state_t, action_t, reward_t in reversed(episode):
            G = reward_t + 0.95 * G
            if state_t not in visited_states:
                visited_states.add(state_t)
                sev, cong = state_t
                returns[(sev, cong)].append(G)
                V[sev, cong] = np.mean(returns[(sev, cong)])
    return V

def run_hospital_mc_control(env, num_episodes=10000, epsilon=0.15, gamma=0.95):
    Q = np.zeros((env.num_severity_states, env.num_congestion_states, len(env.action_space)))
    returns = {(s, c, a): [] for s in range(env.num_severity_states) for c in range(env.num_congestion_states) for a in env.action_space}
    policy = np.zeros((env.num_severity_states, env.num_congestion_states), dtype=int)

    for _ in range(num_episodes):
        state = env.reset()
        episode = []
        done = False

        while not done:
            sev, cong = state
            if random.uniform(0, 1) < epsilon:
                action = random.choice(env.action_space)
            else:
                action = policy[sev, cong]

            next_state, reward, done = env.step(action)
            episode.append((state, action, reward))
            state = next_state

        G = 0
        visited_state_action_pairs = set()
        for state_t, action_t, reward_t in reversed(episode):
            G = reward_t + gamma * G
            state_action = (state_t, action_t)

            if state_action not in visited_state_action_pairs:
                visited_state_action_pairs.add(state_action)
                sev, cong = state_t
                returns[(sev, cong, action_t)].append(G)
                Q[sev, cong, action_t] = np.mean(returns[(sev, cong, action_t)])
                policy[sev, cong] = np.argmax(Q[sev, cong])

    return policy, Q

env = HospitalTriageEnv()

print("Evaluating Baseline Fixed Policy via MC Prediction...")
V_pred = run_hospital_mc_prediction(env)
print(f"Value of critical trauma (Severity: 4) under normal capacity: {V_pred[4, 0]:.2f}")
print(f"Value of critical trauma (Severity: 4) under emergency congestion: {V_pred[4, 1]:.2f}")

print("\nOptimizing Triage Logic via On-Policy MC Control...")
optimal_policy, Q_table = run_hospital_mc_control(env)

print("\n--- Derived Optimal Triage Allocation Matrix ---")
action_labels = ["Standard Care", "Accelerated Triage 🚑", "Critical Overdrive 🚨"]
for s in range(4, 0, -1):
    for c in range(2):
        best_act = optimal_policy[s, c]
        c_status = "Full" if c == 1 else "Normal"
        print(f"Severity Tier: {s} | ER Load: {c_status:<6} -> Strategy: {action_labels[best_act]}")
```
